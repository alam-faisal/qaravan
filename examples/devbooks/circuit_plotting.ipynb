# Circuit plotting

`plot_circuit` accepts any Qaravan `Circuit` and returns the Matplotlib axes.

```python
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
from qaravan.applications.algorithms import trotter_circuit
from qaravan.applications.circuit_library import brickwall_skeleton, ghz_circuit, rxx_layer
from qaravan.core.hamiltonians import TFI
from qaravan.plot import plot_circuit
```

## GHZ preparation

```python
plot_circuit(ghz_circuit(5))
plt.show()
```

## Parameterized brickwall layer

```python
skeleton = brickwall_skeleton(5)
brickwall = rxx_layer(skeleton, params=np.linspace(0.15, 0.75, len(skeleton)))
plot_circuit(brickwall)
plt.show()
```

## One Trotter step

```python
ising_step = trotter_circuit(TFI(5, J=1.0, h=0.6), dt=0.2, order=2)
plot_circuit(ising_step)
plt.show()
```
